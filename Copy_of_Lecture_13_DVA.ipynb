```python
# import required module
import pandas as pd
```

```python
# Converts raw data into a DataFrame

# .head() shows the first 5 rows for verification

df = pd.DataFrame(data)
df.head()
```

```python
# Number of rows and columns
df.shape

# Column names
df.columns

# Summary statistics for numeric columns
df.describe()

# Check for missing values
df.isnull().sum() #count missing data
```

```python
df.isnull().sum()
```

Example 1: Selecting Columns by Exact Names:
Select only the Price, Age, and KM columns from the Toyota dataset.

```python
df.filter(items=['Price', 'Age', 'KM'])
```

Selecting Columns Using like

-> Select all columns that contain the word "Type".

```python
df.filter(like='Type')

# Returns all columns whose names include "Type"

# Useful when working with large datasets
```

Selecting Columns Using regex:
-> Select all columns that start with the letter A.

```python
df.filter(regex='^A')
# ^ means “starts with”, Regex gives maximum flexibility
```

```python
print(df[df['Price'] > 15000])


'''df['Price'] > 15000 creates a Boolean Series (True / False). Pandas keeps only rows where the condition is True.
This is called Boolean Indexing.'''
```

loc accessor: The .loc[] method allows for more complex filtering, used to filter both rows and columns at the same time by specifying conditions for both axes. It allows to specify conditions directly within the square brackets.

```python
df.loc[df['Price'] > 15000, ['Price', 'FuelType']]

# .loc[] is a label-based indexer. First part filters rows. Second part selects specific columns.
```

Multi-Condition Filtering:
Each condition produces a Boolean Series.

& combines conditions element-wise.

Parentheses ensure correct order of evaluation.

Only rows where ALL conditions are true are returned.

Models real-world decision logic.

```python
df[
    (df['FuelType'] == 'Petrol') &
    (df['Age'] > 10) &
    (df['Price'] < 10000)
]
```

Using .isin():

.isin() checks whether each value belongs to a given list.

Returns True if value is 3 OR 5.

Cleaner and safer than writing multiple OR conditions.

Used for category-based filtering.

```python
df[df['Doors'].isin([3, 5])]
```

Grouping: Summarizing Data
The groupby() method allows you to group your data and execute functions on these groups.

# GroupBy + Single Aggregation
*  Split data by FuelType
*  Apply mean on Price
*   Combine results into one output

```python
df.groupby('FuelType')['Price'].mean()
```

Multiple Metrics using .agg()

*   .agg() allows multiple calculations at once
*   Column renaming makes output readable
*   Used heavily in business reporting

```python
df.groupby(['FuelType', 'Doors'])['Price'].mean()
```
