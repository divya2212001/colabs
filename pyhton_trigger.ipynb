```python
def trigger_function():
    print("Triggered! Email received from divyanjali.gopisetty@fireai.in")
```

```python
def send_reply(recipient_email):
    try:
        server = smtplib.SMTP('smtp.gmail.com', 587)
        server.starttls()
        server.login(EMAIL, APP_PASSWORD)

        reply_body = "automated response from sender"
        msg = MIMEText(reply_body)
        msg['Subject'] = 'Auto-Reply'
        msg['From'] = EMAIL
        msg['To'] = recipient_email

        server.send_message(msg)
        server.quit()
        print(f"Success: Auto-reply sent back to {recipient_email}")

    except Exception as e:
        print(f"Error sending reply: {e}")
```

```python
def monitor_mailbox():
    mail = imaplib.IMAP4_SSL("imap.gmail.com")
    mail.login(EMAIL, APP_PASSWORD)
    mail.select("INBOX")

    processed = set()

    while True:
        mail.noop()
        status, messages = mail.search(None, 'UNSEEN')

        if status == "OK":
            for email_id in messages[0].split():

                if email_id in processed:
                    continue

                status, msg_data = mail.fetch(email_id, "(RFC822)")

                for response in msg_data:
                    if isinstance(response, tuple):
                        msg = email.message_from_bytes(response[1])

                        sender = msg["From"]

                        if "divyanjali.gopisetty@fireai.in" in sender:
                            trigger_function()
                            send_reply(sender)

                processed.add(email_id)

        time.sleep(2)

monitor_mailbox()
```
