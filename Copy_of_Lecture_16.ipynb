**What is Univariate Analysis?**

**Univariate Analysis** is the **simplest form of data analysis**. The term "Uni" means "one," implying that we are analyzing one variable at a time. Its **primary purpose** is to **describe the data and find patterns within a single column** without looking at causes or relationships between variables.

**Data Types and Plot Selection**
The type of plot you choose depends on the Data Type of the variable:

**1.Numerical (Continuous):** Data that can take any value in a range (e.g., Sales or Profit).

**2.Numerical (Discrete):** Countable data (e.g., Quantity of items).

**3.Categorical:** Data that falls into specific groups or labels (e.g., State, Category, or Gender).

We use frequencies to understand how categories are distributed.

**Activity:**
Calculate the frequency of product categories. The source data contains categories like "Binders," "Paper," "Furnishings," and "Phones"

```python
# Get absolute frequency of the 'Category' column
abs_freq = df['Category'].value_counts()
print("Absolute Frequencies:\n", abs_freq)
```

#### Activity 1: Running relative frequency

```python
rel_freq = df['Category'].value_counts(normalize=True)
print("\nRelative Frequency (Proportion):\n", rel_freq)
```

```python
# Calculate Relative Frequency (percentage)
rel_freq = df['Category'].value_counts(normalize=True) * 100
print("\nRelative Frequencies (%):\n", rel_freq)
```

### Checkpoint: Percentage of orders in 'Binders'

```python
binder_pct = rel_freq.get('Binders', 0)
print(f"\nPercentage of orders in 'Binders': {binder_pct:.2f}%")
```

**Visualization: Top States and The Country Variable**

Sorting bars in a bar chart is a vital design principle because it allows for immediate comparison of the largest and smallest values. We will also look at the Country variable; since all entries in this dataset are from the "United States", a Pie Chart would simply show one solid block, highlighting a data issue where a variable lacks variance.

```python
import seaborn as sns
import matplotlib.pyplot as plt
# Create a Bar Chart for Top 5 States
plt.figure(figsize=(10, 6))
top_5_states = df['State'].value_counts().head(5)

# Sorting the bars for better readability
sns.barplot(x=top_5_states.index, y=top_5_states.values, palette='viridis',
            order=top_5_states.index) # Sorting is handled by the value_counts order

plt.title('Top 5 States by Order Volume (Sorted)')
plt.xlabel('State')
plt.ylabel('Number of Orders')
plt.show()
```

**The Pareto Principle (80/20 Rule)**

The 80/20 Rule suggests that roughly 80% of effects come from 20% of causes. In retail, this often means 80% of sales come from just 20% of states or categories. By grouping our sales data by state, we can identify these key drivers.

```python
# Group by State and sum the Sales, then sort descendingly
state_sales = df.groupby('State')['Sales'].sum().sort_values(ascending=False)
# total_sales = df["Sales"].sum()
# print("Top 5 States accounting for major Sales:")
# print(state_sales.head(5)/total_sales*100)

top_5_sales = state_sales.head(5)

plt.figure(figsize=(8, 8))

plt.pie(
    top_5_sales.values,
    labels=top_5_sales.index,
    autopct='%1.1f%%',
    startangle= 140
)

plt.title("Top 5 States Contribution to Total Sales")
plt.axis('equal')  # Ensures pie is a circle
plt.show()
```

**Numerical Analysis: Histograms and Distribution**

A Histogram is the primary tool for univariate analysis of continuous numerical data. The bin size is crucial: if bins are too wide, you hide the distribution details; if they are too narrow, the chart becomes "noisy." We will plot the distribution of Profit, which ranges from significant losses to high gains in the sources.

**Design Focus:** We must use clear titles, label axes properly, and choose color palettes that are accessible

```python
# Filter California and Plot Sales Histogram
california_data = df[df['State'] == 'California']

plt.figure(figsize=(10, 6))
sns.histplot(california_data['Sales'], bins=20,stat='density', color='orange', kde=True)
sns.kdeplot(california_data['Sales'],color='red', linewidth=2)
plt.title('Distribution of Sales in California')
plt.xlabel('Sales Amount ($)')
plt.show()
```

**Advanced Technique: Small Multiples**

When comparing distributions across categories, plotting them all on one axis creates overplotting (a mess of overlapping lines).

Instead, we use Small Multiples to create separate, side-by-side histograms. This aids comparison while maintaining clarity.

```python
# Frequency Distribution of Product Categories
df['Category'].value_counts().plot(kind='barh', figsize=(8,6),color='skyblue')
plt.title('Frequency of Product Categories')
plt.xlabel('Count')
plt.ylabel('Category')
plt.show()
```
